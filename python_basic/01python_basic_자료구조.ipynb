## 1. 리스트(list) — 순서가 있는 값의 모음

대괄호 `[]` 로 만들며, 순서(인덱스)가 있고 값을 바꿀 수 있습니다(가변).

```python
# 리스트 생성
fruits = ['사과', '바나나', '체리']
print(fruits)
print('개수:', len(fruits))  # len(): 요소 개수

# 인덱싱: 0부터 시작, 음수는 뒤에서부터
print('첫번째:', fruits[0])
print('마지막:', fruits[-1])
```

```python
# 슬라이싱: [시작:끝]  (끝 인덱스는 포함하지 않음)
numbers = [10, 20, 30, 40, 50]
print(numbers[1:3])   # 인덱스 1,2 → [20, 30]
print(numbers[:2])    # 처음부터 인덱스 1까지 → [10, 20]
print(numbers[2:])    # 인덱스 2부터 끝까지 → [30, 40, 50]

# 참고: 스크래핑 코드의 a_tags[0].click() 처럼 '첫번째 요소'를 자주 사용합니다.
```

```python
# 리스트 수정: 추가/삭제
fruits.append('포도')       # 맨 뒤에 추가
print('추가 후:', fruits)

fruits.remove('바나나')     # 값으로 삭제
print('삭제 후:', fruits)

# 반복문으로 순회 (스크래핑에서 결과 목록을 돌 때 사용)
for fruit in fruits:
    print('-', fruit)
```

## 2. 튜플(tuple) — 바꿀 수 없는 값의 모음

소괄호 `()` 로 만들며, 한 번 만들면 값을 바꿀 수 없습니다(불변).
함수가 여러 값을 반환할 때 자주 쓰입니다.

```python
# 튜플 생성
point = (3, 5)
print(point[0], point[1])

# 언패킹(unpacking): 튜플의 값을 여러 변수에 한 번에 나눠 담기
x, y = point
print('x =', x, ', y =', y)

# 앞서 만든 자격증명 함수가 'return username, secret' 로 튜플을 돌려주고
# 'GIT_USERNAME, GIT_PASSWORD = load_...()' 로 언패킹하는 것과 같은 원리입니다.
```

## 3. 딕셔너리(dict) — 키:값 쌍의 모음 ⭐

중괄호 `{}` 로 만들며 `키(key)` 로 `값(value)` 을 찾습니다.
**API 응답(JSON)의 한 건이 바로 딕셔너리** 이므로 가장 중요합니다.

```python
# 딕셔너리 생성 (도서 한 권을 표현)
book = {
    'title': '파이썬 입문',
    'author': '홍길동',
    'price': 25000,
}

# 값 꺼내기: 키로 접근
print(book['title'])

# get(): 키가 없어도 에러 대신 기본값 반환 (안전)
print(book.get('publisher', '출판사 미상'))

# 값 추가/수정
book['discount'] = 22000
print(book)
```

```python
# 딕셔너리 순회
for key in book.keys():        # 키만
    print('키:', key)

print('-' * 20)

for key, value in book.items():  # 키와 값 함께 (가장 많이 사용)
    print(f'{key} = {value}')
```

## 4. 셋(set) — 중복 없는 모음

중괄호 `{}` 로 만들며 중복을 자동 제거합니다. 순서는 없습니다.

```python
# 중복된 값이 자동으로 하나만 남는다
genres = ['발라드', '댄스', '발라드', '힙합', '댄스']
unique_genres = set(genres)
print(unique_genres)
print('고유 장르 수:', len(unique_genres))

# 활용: 리스트에서 중복 제거할 때 set() 으로 감쌌다가 다시 list() 로 변환
print(list(unique_genres))
```

## 5. 중첩 구조 — JSON과 같은 형태 ⭐

실제 API 응답은 **딕셔너리 안에 리스트, 리스트 안에 딕셔너리** 가 섞인 중첩 구조입니다.
이 구조를 읽을 수 있어야 스크래핑 결과를 다룰 수 있습니다.

```python
# '딕셔너리들의 리스트' — pandas.DataFrame 으로 바로 변환되는 대표 형태
books = [
    {'title': '파이썬 입문', 'price': 25000},
    {'title': '데이터 분석', 'price': 30000},
    {'title': '웹 스크래핑', 'price': 28000},
]

# 두번째 책의 제목 꺼내기: 리스트 인덱스 → 딕셔너리 키
print(books[1]['title'])

# 전체 순회하며 제목만 출력
for b in books:
    print(b['title'], '-', b['price'], '원')
```

```python
# 딕셔너리 안에 리스트가 들어있는 형태 (검색 API 응답과 유사)
response = {
    'query': '파이썬',
    'total': 3,
    'items': books,   # 위에서 만든 리스트를 값으로
}

print('검색어:', response['query'])
print('결과 수:', response['total'])

# response['items'] 는 리스트 → 첫 항목은 딕셔너리
print('첫 결과 제목:', response['items'][0]['title'])

# 실제 코드의 res.json().get('items', []) 가 바로 이 구조에서 items 리스트를 꺼내는 것입니다.
```

## 정리
- **리스트** `[]`: 순서 있음, 수정 가능 → 여러 건의 결과 목록
- **튜플** `()`: 수정 불가 → 함수의 다중 반환값, 언패킹
- **딕셔너리** `{}`: 키로 값 조회 → JSON 한 건
- **셋** `set()`: 중복 제거
- **중첩 구조**: '딕셔너리들의 리스트' = 표(DataFrame)의 원형
