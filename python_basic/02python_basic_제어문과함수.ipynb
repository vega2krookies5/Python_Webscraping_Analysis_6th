## 1. 조건문 if / elif / else

파이썬은 **들여쓰기(보통 공백 4칸)** 로 코드 블록을 구분합니다.

```python
price = 28000

if price >= 30000:
    print('비쌈')
elif price >= 20000:
    print('보통')
else:
    print('저렴')

# 스크래핑 코드의 'if a_tags:' 처럼, 비어있지 않은 리스트는 True 로 취급됩니다.
results = []
if results:
    print('결과 있음')
else:
    print('결과 없음 (빈 리스트는 False)')
```

## 2. 반복문 for / while

```python
# for: 리스트/범위를 순회
for name in ['BTS', 'IU', 'NCT']:
    print(name)

print('-' * 20)

# range(1, 4) → 1, 2, 3 (끝 숫자는 포함 안 함)
# 스크래핑에서 page-1, page-2 ... 순회할 때 사용
for page in range(1, 4):
    print(f'{page} 페이지 요청')
```

```python
# enumerate(): 인덱스와 값을 함께
for i, name in enumerate(['BTS', 'IU', 'NCT']):
    print(i, name)

print('-' * 20)

# while: 조건이 True인 동안 반복
count = 0
while count < 3:
    print('count =', count)
    count += 1   # count = count + 1
```

```python
# break: 반복 중단 / continue: 다음 반복으로 건너뛰기
for n in range(1, 10):
    if n == 5:
        break        # 5에서 멈춤
    if n % 2 == 0:
        continue     # 짝수는 건너뜀
    print(n)         # 1, 3 만 출력
```

## 3. 리스트 컴프리헨션 — 반복을 한 줄로

`[식 for 항목 in 반복대상 if 조건]` 형태로, 리스트를 간결하게 만듭니다.
실제 코드의 `[c for c in df.columns if c not in [...]]` 가 이 문법입니다.

```python
# 일반 반복문
squares = []
for n in range(1, 6):
    squares.append(n * n)
print('반복문:', squares)

# 같은 결과를 컴프리헨션으로
squares2 = [n * n for n in range(1, 6)]
print('컴프리헨션:', squares2)

# if 조건 추가: 짝수의 제곱만
even_squares = [n * n for n in range(1, 11) if n % 2 == 0]
print('짝수 제곱:', even_squares)
```

## 4. 함수 정의와 반환

`def 함수명(매개변수):` 로 정의하고 `return` 으로 결과를 돌려줍니다.
함수 첫 줄의 `"""설명"""`(docstring)으로 용도를 적어두면 좋습니다.

```python
def add(a, b):
    """두 수를 더해 반환한다."""
    return a + b

print(add(3, 5))


def swap(a, b):
    """두 값의 순서를 바꿔 튜플로 반환한다."""
    return b, a   # 여러 값을 반환하면 튜플

# 언패킹으로 받기
x, y = swap(10, 20)
print('x =', x, ', y =', y)
```

## 5. 기본값 인자 / 키워드 인자

실제 코드의 `def scrape_books(max_pages=3, delay=1.0):` 처럼
매개변수에 기본값을 주면 호출 시 생략할 수 있습니다.

```python
def greet(name, greeting='안녕하세요'):
    """인사말을 만든다. greeting 은 기본값이 있어 생략 가능."""
    return f'{greeting}, {name}님!'

print(greet('철수'))                       # 기본값 사용
print(greet('영희', '반가워요'))            # 위치 인자로 전달
print(greet('민수', greeting='환영합니다'))  # 키워드 인자로 전달 (이름 지정)
```

## 6. 람다(lambda) — 이름 없는 한 줄 함수

간단한 함수를 즉석에서 만들 때 사용합니다. 숫자 천단위 콤마 포맷 등에 자주 쓰입니다.

```python
# 일반 함수
def double(x):
    return x * 2

# 같은 것을 람다로
double2 = lambda x: x * 2
print(double(5), double2(5))

# 천단위 콤마 포맷 (숫자 → '1,234,567')
comma = lambda val: '{:,}'.format(int(val))
print(comma('1234567'))

# 정렬 기준으로 람다 사용 (가격 높은 순)
books = [{'title': 'A', 'price': 25000}, {'title': 'B', 'price': 30000}]
books_sorted = sorted(books, key=lambda b: b['price'], reverse=True)
print([b['title'] for b in books_sorted])
```

## 정리
- **if/elif/else**: 조건 분기 (빈 리스트/None 은 False)
- **for/while + range/enumerate**: 반복 (페이지 순회 등)
- **컴프리헨션**: `[식 for x in 대상 if 조건]` 으로 간결하게
- **함수**: `def` + `return`, 기본값·키워드 인자
- **lambda**: 즉석 한 줄 함수 (정렬 key, 포맷 등)
