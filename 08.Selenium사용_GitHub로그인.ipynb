### WebDriver 의 Chrome 객체생성
* python.org 웹사이트로 요청을 보내기
* pypi 메뉴를 클릭해서 pypi.org 웹사이트로 이동하기

````python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get("https://www.python.org/")

#webdriver에 포함된 Chrome 객체를 생성 - chromedriver.exe 사용해서

#li 엘리먼트 선택해서 가져오기
li_tags = driver.find_elements(By.CSS_SELECTOR,'div#top ul.menu li')
print(li_tags)
#pypi 엘리먼트를 찾아서 저장할 변수 선언
pypi = None
for li_tag in li_tags:
    print(li_tag.text, type(li_tag))
    if li_tag.text == 'PyPI':
        pypi = li_tag
        
time.sleep(3)

#pypi 엘리멘트의 click() 함수 호출
pypi.click()

#5초로 waiting time 설정
time.sleep(5)
#driver 종료
driver.quit()
````

### 로그인 실패할 경우 처리한 코드

````python
from dotenv import load_dotenv
import os

load_dotenv()

GIT_USERNAME = os.getenv("GIT_USERNAME")
print(GIT_USERNAME[:4])

GIT_PASSWORD = os.getenv("GIT_PASSWORD")
````

````python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

from selenium.common.exceptions import NoSuchElementException, TimeoutException
import time

# 단계 사이에 잠시 멈춰 화면 전환을 눈으로 확인하기 위한 대기 시간(초)
# 값을 늘리면 더 천천히, 0 으로 하면 대기 없이 빠르게 동작한다.
STEP_PAUSE = 3

url = 'https://github.com/login'

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
# 명시적 대기 객체: 최대 10초까지 조건이 만족될 때까지 기다린다.
wait = WebDriverWait(driver, 10)

# github login 페이지 요청
driver.get(url)

# 로그인 입력창이 나타날 때까지 기다린 뒤 아이디/비밀번호 입력
wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'input#login_field')))
driver.find_element(By.CSS_SELECTOR, 'input#login_field').send_keys(GIT_USERNAME)
driver.find_element(By.CSS_SELECTOR, 'input#password').send_keys(GIT_PASSWORD)
time.sleep(1)

# 비밀번호 필드에서 엔터 → 로그인 시도
driver.find_element(By.CSS_SELECTOR, 'input#password').send_keys(Keys.RETURN)
time.sleep(2)

# 로그인 성공 여부 판단 (실패 시에만 오류 배너 div.js-flash-alert 존재)
login_ok = None
try:
    div_error = driver.find_element(By.CSS_SELECTOR, 'div.js-flash-alert')
    if div_error:
        login_ok = False
        print("login fail ----------------", login_ok)
except NoSuchElementException:
    login_ok = True
except Exception as exp:
    print(exp)

print('login_ok', login_ok)

if login_ok:
    try:
        # 로그인된 홈 화면을 잠시 보여준다
        print("로그인 성공 - 홈 화면 확인")
        time.sleep(STEP_PAUSE)

        # 1) 저장소 목록 페이지로 '직접 이동'
        #    아바타 드롭다운(GitHub UI 변경에 취약)을 거치지 않아 훨씬 안정적이다.
        driver.get(f"https://github.com/{GIT_USERNAME}?tab=repositories")
        print("저장소 목록으로 이동")
        time.sleep(STEP_PAUSE)  # 목록 화면을 눈으로 확인

        # 2) 저장소 목록에서 첫번째 저장소 클릭
        first_repo = wait.until(EC.element_to_be_clickable(
            (By.CSS_SELECTOR,
             "#user-repositories-list h3 a, li.col-12 h3.wb-break-all a")))
        repo_name = first_repo.text
        print("첫번째 저장소:", repo_name)
        time.sleep(STEP_PAUSE)  # 클릭 직전에 잠시 멈춰 어떤 저장소인지 확인
        first_repo.click()

        # 저장소 페이지가 로딩될 때까지 대기 (저장소 헤더 요소 확인)
        wait.until(EC.presence_of_element_located(
            (By.CSS_SELECTOR, "#repository-container-header, strong[itemprop='name']")))
        print("현재 URL:", driver.current_url)
        time.sleep(STEP_PAUSE)  # 저장소 화면을 눈으로 확인

        # 3) 로그아웃
        #    GitHub 로그아웃은 /logout 으로 POST 하는 <form> 이다.
        #    (드롭다운의 'Sign out' 은 링크가 아니라 form 안의 버튼)
        #    따라서 form 을 직접 submit 하는 것이 가장 안정적이다.
        logout_forms = driver.find_elements(By.CSS_SELECTOR, "form[action='/logout']")
        if logout_forms:
            logout_forms[0].submit()  # find_elements 는 리스트를 반환하므로 [0]
            # 로그아웃 후 로그인 링크/폼이 다시 나타나면 성공
            wait.until(EC.presence_of_element_located(
                (By.CSS_SELECTOR, "a[href='/login'], input#login_field")))
            print("로그아웃 완료:", driver.current_url)
            time.sleep(STEP_PAUSE)  # 로그아웃된 화면을 눈으로 확인
        else:
            print("로그아웃 form 을 찾지 못했습니다. (GitHub UI 변경 가능성)")

    except TimeoutException as texp:
        print("요소 대기 시간 초과 - 셀렉터를 현재 GitHub DOM 에 맞게 확인하세요:", texp)
    except Exception as exp:
        print("에러:", exp)

time.sleep(3)
driver.quit()
````

### 자격 증명을 안전하게 관리하기 (평문 .env 대체)

`.env` 에 비밀번호를 평문으로 두는 대신 다음을 사용한다.

1. **`getpass`** — 실행 시 입력받되 화면/출력에 노출되지 않음 (어디에도 저장 안 함)
2. **`keyring`** — 입력한 값을 OS 보안 저장소(Windows 자격 증명 관리자)에 **암호화 저장** 후 재사용
3. **PAT(Personal Access Token)** — 비밀번호 대신 사용 권장 (권한 최소화·만료·폐기 가능)

> 설치: `pip install keyring`  
> 아래 셀을 실행하면 `GIT_USERNAME` / `GIT_PASSWORD` 변수가 채워지므로, 위 로그인 셀에서 그대로 사용할 수 있다. (평문 .env 를 읽는 cell 대신 이 셀을 사용)

#### 1. GUI — 자격 증명 관리자
control /name Microsoft.CredentialManager
→ Windows 자격 증명 탭 → 일반 자격 증명 목록에서 github-selenium-demo 검색
(여기서 "표시" 를 누르면 Windows 계정 암호 확인 후 값도 볼 수 있습니다)

#### 2. CLI — cmdkey (목록만, 비밀번호는 안 보임) - powershell 에서 확인 가능
cmdkey /list | Select-String "github-selenium-demo" -Context 0,2

#### 3. Python — 값 자체 확인
```
import keyring
print(keyring.get_keyring())                                   # 백엔드 확인
print(keyring.get_password("github-selenium-demo", "myjungwu")) # 저장된 값 (출력 주의!)
```
삭제는 노트북 마지막 셀의 주석 부분(keyring.delete_password(...)) 또는 GUI에서 "제거" 로 가능합니다.

````python
# =========================================================
# 자격 증명을 안전하게 가져오기
#  - 평문 .env 대신 getpass 입력 + keyring(OS 보안 저장소) 사용
#  - 비밀번호 대신 GitHub Personal Access Token(PAT) 권장
# =========================================================
import os
import getpass

# keyring: OS 보안 저장소(Windows 자격 증명 관리자 등)에 암호화 저장
# 설치: pip install keyring
try:
    import keyring
    HAS_KEYRING = True
except ImportError:
    HAS_KEYRING = False
    print("keyring 미설치 → getpass 입력만 사용합니다. (pip install keyring 권장)")

# 보안 저장소에서 이 자격 증명을 구분하는 이름
SERVICE_NAME = "github-selenium-demo"


def load_github_credentials(service=SERVICE_NAME):
    """GitHub 자격 증명을 안전하게 가져온다.

    우선순위:
      1) OS 보안 저장소(keyring)에 저장돼 있으면 그대로 사용
      2) 없으면 getpass 로 입력받고(화면에 노출되지 않음),
         keyring 이 있으면 보안 저장소에 저장해 다음부터 재사용
    Returns:
        (username, secret) 튜플
    """
    # 사용자명은 민감정보가 아니므로 환경변수 또는 입력으로 받는다
    username = os.getenv("GIT_USERNAME") or input("GitHub 사용자명: ").strip()

    secret = keyring.get_password(service, username) if HAS_KEYRING else None

    if not secret:
        # getpass: 입력값이 노트북 출력/화면에 남지 않는다
        secret = getpass.getpass(f"'{username}' 의 GitHub 비밀번호 또는 PAT 입력: ")
        if HAS_KEYRING:
            keyring.set_password(service, username, secret)
            print("→ OS 보안 저장소에 저장 완료. 다음 실행부터는 입력이 필요 없습니다.")
    else:
        print("→ OS 보안 저장소에서 자격 증명을 불러왔습니다.")

    return username, secret


# 실행: 이후 로그인 셀의 GIT_USERNAME / GIT_PASSWORD 로 사용
GIT_USERNAME, GIT_PASSWORD = load_github_credentials()

# 값 자체는 절대 출력하지 않는다 (길이/일부만 확인)
print("사용자명:", GIT_USERNAME)
print("비밀번호 확보됨, 길이:", len(GIT_PASSWORD))

# 저장된 자격 증명을 삭제하려면 아래 주석을 해제해 실행:
# if HAS_KEYRING:
#     keyring.delete_password(SERVICE_NAME, GIT_USERNAME)
#     print("보안 저장소에서 삭제했습니다.")
````

````python
# =========================================================
# 비밀값 확인용 마스킹 유틸
#  - print(keyring.get_password(...)) 처럼 값을 그대로 찍으면
#    그 출력이 .ipynb 파일에 그대로 저장되어 git 에 커밋된다
#  - 확인이 필요할 때는 반드시 이 함수를 거친다
# =========================================================
def mask(secret, keep=2):
    """비밀값을 앞 keep 자리만 남기고 마스킹한다.

    Args:
        secret (str): 마스킹할 문자열
        keep (int): 앞에서 남길 글자 수

    Returns:
        str: 'gh****************** (len=20)' 형태의 표시용 문자열
    """
    if not secret:
        return "(없음)"
    return f"{secret[:keep]}{'*' * (len(secret) - keep)} (len={len(secret)})"


# 사용 예
print(mask("ghp_1234567890abcdef"))
print(mask(""))
````

### 비밀값을 화면에 안전하게 표시하는 3가지 방법

`keyring` 에서 꺼낸 값을 확인하고 싶을 때, **인코딩 → 출력 → 디코딩 → 사용** 은 정답이 아니다.

| 방법 | 되돌리기 | 노트북에 저장돼도 안전? | 용도 |
|---|---|---|---|
| `base64` 인코딩 | ✅ 키 없이 누구나 | ❌ **평문과 동일** | 어깨너머 훔쳐보기 방어뿐 |
| SHA-256 **지문** | ❌ 불가능 | ✅ 안전 | "저장된 값이 맞나" 확인 |
| **`SecretStr` 래퍼** | ✅ `.get()` 호출 시 | ✅ 안전 | **출력은 막고, 쓸 때는 원본** |

> **핵심**: 인코딩(encoding)은 암호화(encryption)가 아니다.
> **키가 필요 없으면 보호 장치가 아니다.**
>
> "출력할 땐 가리고, 쓸 땐 원본"의 정석 답은 **3번째 방법 `SecretStr`** 이다.

````python
# =========================================================
# 방법 1) base64 인코딩 - "되돌릴 수 있지만, 보안 장치는 아니다"
# ---------------------------------------------------------
# base64 는 '암호화(encryption)'가 아니라 '표현 방식 변환(encoding)'이다.
#   - 암호화 : 키가 있어야만 복원 가능      → 키가 없으면 읽을 수 없다
#   - 인코딩 : 키 없이 누구나 복원 가능     → 사실상 평문과 동일하다
#
# 원래 목적은 '바이너리 데이터를 텍스트로만 전송하는 통로(메일, JSON 등)에
# 안전하게 실어 보내기' 위한 것이지, 값을 숨기기 위한 것이 아니다.
#
# 따라서 인코딩된 값이 .ipynb 에 저장되어 git 에 커밋되면
# 평문을 커밋한 것과 보안상 아무 차이가 없다.
# 인코딩/디코딩 Online Site https://www.base64encode.net/
# =========================================================
import base64

demo_pwd = "hunter2_secret!"  # 데모용 가짜 비밀번호 (실제 값은 절대 넣지 말 것)

# --- 인코딩 : str → bytes → base64 bytes → str ---
#   .encode()    : 문자열을 UTF-8 바이트로 변환 (base64 는 bytes 만 처리)
#   b64encode()  : 바이트를 base64 바이트로 변환
#   .decode()    : 결과 바이트를 다시 문자열로 (print 하기 위해)
print(".encode() :", demo_pwd.encode())
print("b64encode() :", base64.b64encode(demo_pwd.encode()))

encoded = base64.b64encode(demo_pwd.encode()).decode()
print("인코딩 결과:", encoded)
print()
# --- 디코딩 : 위 과정을 그대로 역순으로 ---
# 키(비밀번호)가 전혀 필요 없다는 점에 주목할 것
decoded = base64.b64decode(encoded).decode()
print("디코딩 결과:", decoded)

print()
print("→ 키 없이 한 줄로 복원됐다. 노트북에 저장되면 평문 유출과 동일하다.")
````

````python
# =========================================================
# 방법 2) SHA-256 지문 - "되돌릴 수 없지만 확인은 된다"
# ---------------------------------------------------------
# 해시(hash)는 '단방향 함수'다.
#   원본 → 지문 : 계산할 수 있다
#   지문 → 원본 : 계산할 수 없다  ← 이 점이 base64 와 결정적으로 다르다
# 따라서 지문은 노트북에 저장되어 git 에 커밋돼도 안전하다.
#
# 어디에 쓰나?
#   "keyring 에 저장된 값이 내가 방금 입력한 그 값이 맞나?" 처럼
#   값을 드러내지 않고 '같은 값인지'만 비교할 때 쓴다.
#   len() 으로 길이만 보는 것보다 훨씬 정확하다.
#   (길이가 같아도 값이 다를 수 있지만, 지문이 같으면 값도 같다)
# =========================================================
import hashlib

def fingerprint(secret, n=8):
    """비밀값의 SHA-256 지문 앞 n자리를 돌려준다.

    Args:
        secret (str): 지문을 만들 원본 문자열
        n (int): 표시할 16진수 자릿수 (8자리 = 32비트)

    Returns:
        str: 16진수 지문 문자열. 원본으로 되돌릴 수 없다
    """
    if not secret:
        return "(없음)"
    # encode()    : 문자열 → UTF-8 바이트 (해시 함수는 bytes 만 처리)
    # sha256()    : 해시 계산
    # hexdigest() : 결과를 16진수 문자열로 (64자리) → 앞 n자리만 사용
    return hashlib.sha256(secret.encode()).hexdigest()[:n]


print("원본        :", fingerprint("hunter2_secret!"))
print("같은 값     :", fingerprint("hunter2_secret!"))  # 항상 같은 지문
print("한 글자 차이 :", fingerprint("hunter2_secret?"))  # 완전히 다른 지문

print()
print("→ 같은 값이면 지문도 같다 = 동일성 확인 가능")
print("→ 한 글자만 달라도 지문이 완전히 달라진다 (눈사태 효과, avalanche effect)")
print("→ 지문만으로는 원본을 알 수 없으므로 노트북에 저장돼도 안전하다")
````

````python
# =========================================================
# 방법 3) SecretStr 래퍼 - "출력은 막고, 쓸 때는 원본" ★ 권장
# ---------------------------------------------------------
# 앞의 두 방법은 '내가 조심해서 마스킹 함수를 호출할 때'만 안전하다.
# 실수로 print(GIT_PASSWORD) 를 한 번 치면 그대로 노출된다.
#
# SecretStr 은 값을 객체로 감싸서, 파이썬이 그 객체를 '문자열로 표현할 때'
# 항상 마스킹된 형태만 내보내도록 만든다.
#   __str__  : print(obj) 와 f"{obj}" 가 호출한다
#   __repr__ : 셀 마지막 줄에 obj 만 썼을 때,
#              그리고 리스트/딕셔너리 안에 들어있을 때 호출된다
# → 두 개를 모두 덮어야 새는 구멍이 없다.
#
# 원본은 .get() 을 명시적으로 호출할 때만 나온다.
# → 코드 전체에서 '.get()' 만 검색하면 노출 지점을 전부 찾을 수 있다.
#
# 참고: pydantic 라이브러리의 SecretStr 이 이와 같은 방식이다.
#       (pip install pydantic → from pydantic import SecretStr)
# =========================================================
import hashlib

class SecretStr:
    """비밀값을 감싸 실수로 출력되는 것을 막는 래퍼 클래스.

    Examples:
        >>> pwd = SecretStr("hunter2")
        >>> print(pwd)
        SecretStr(***, len=7)
        >>> pwd.get()
        'hunter2'
    """

    # __slots__ : 인스턴스가 가질 수 있는 속성을 _value 하나로 제한한다.
    #             메모리를 아끼고, 다른 이름으로 값이 복사되는 실수를 줄인다.
    __slots__ = ("_value",)

    def __init__(self, value):
        """
        Args:
            value (str): 감쌀 비밀 문자열. None 이면 빈 문자열로 처리
        """
        self._value = value or ""

    def get(self):
        """실제 비밀값을 꺼낸다.

        Returns:
            str: 원본 문자열. 꼭 필요한 순간에만 호출할 것
        """
        return self._value

    def __repr__(self):
        """객체를 표현할 때 항상 마스킹된 문자열을 돌려준다.

        셀 마지막 줄에 변수만 쓰거나, 리스트/딕셔너리에 담겨 출력될 때 호출된다.
        """
        return f"SecretStr(***, len={len(self._value)})"

    # print() 와 f-string 도 같은 결과를 쓰도록 __str__ 을 __repr__ 에 연결한다.
    # 이 한 줄이 없으면 print(obj) 가 기본 __str__ 을 타서 구멍이 생길 수 있다.
    __str__ = __repr__

    def __len__(self):
        """len(pwd) 로 값을 드러내지 않고 길이만 확인할 수 있게 한다."""
        return len(self._value)

    def __bool__(self):
        """if pwd: 형태로 값이 비어있는지 검사할 수 있게 한다."""
        return bool(self._value)

    def fingerprint(self, n=8):
        """값을 드러내지 않고 동일성만 비교할 수 있는 지문을 돌려준다.

        Args:
            n (int): 표시할 16진수 자릿수

        Returns:
            str: SHA-256 지문 앞 n자리
        """
        if not self._value:
            return "(없음)"
        return hashlib.sha256(self._value.encode()).hexdigest()[:n]


print("SecretStr 클래스 정의 완료")
````
